Write a program that asks the user for a number of days. The program then
prints out the number of seconds in the amount of days given.

```python
days = int(input("Enter the number of days: "))
seconds_in_a_day = 60*60*24# 60 seconds in a minute and 60 * 60 seconds in one hour then 60 * 60 *24 seconds in one day which has 24 hours
print(days," days have a total of ",seconds_in_a_day*days," seconds.")
```

Write a program that asks a user to input the radius then the program
calculates the volume of a sphere.

```python
import math

# Access the value of Pi
pi = math.pi
r = int(input("Input the radius of the sphere: "))
volume = (4/3)*pi*(r**3)# volume is 4/3*pi*radius*radius*radius
print("The volume of a sphere of radius ",r ," is: ",volume, " units cubed.")
```

Program that store users input in an array then calculates the average

```python
# This program prompts the user to input 5 numbers

def calculate_average():

    numbers = []

    # for loop to get 5 values from the user.

    for i in range(5):
        while True:
            try:
                user_input = input(f"Enter number {i + 1} of 5: ")

                # Convert the input string to a floating-point number which handles whole numbers and decimals
                number = float(user_input)

                # Adding the converted number to the list.
                numbers.append(number)
                break  # Exit the inner while loop after a valid number is entered.
            except ValueError:
                print("Invalid input. Please enter a number.")


    if numbers:

        total_sum = sum(numbers)

        # Average by dividing the sum by the number of elements in the list.
        average = total_sum / len(numbers)

        # Display the calculated average to the user.
        print(f"\nNumbers entered: {numbers}")
        print(f"The average of the numbers is: {average}")
    else:
        print("No numbers were entered.")

calculate_average()#The function Call
```
